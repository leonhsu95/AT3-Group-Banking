### Helper Functions

```python
import numpy as np

UNIT = 'Records'
BAR_COLOR = '#2563eb'
AXIS_CFG = dict(domainColor='#9ca3af', tickColor='#9ca3af', labelColor='#000', titleColor='#000',
                labelFontSize=12, titleFontSize=13, labelLimit=160)


def _finish(chart, title, width, height):
    """Apply the title, size and axis styling shared by every chart in this notebook."""
    return (chart.properties(title=alt.TitleParams(title, fontSize=15, anchor='start'), width=width, height=height)
                 .configure_view(strokeWidth=0)
                 .configure_axis(**AXIS_CFG))


def _bar_chart(data, category_field, value_field, *, category_title, value_title,
               sort_order, tooltip, color, horizontal, title, height=None):
    """One bar per category, laid out as columns, or as rows when horizontal=True. Width and height
    are capped so the chart always fits on screen without scrolling."""
    if horizontal:
        encoding = dict(
            y=alt.Y(f'{category_field}:N', title=category_title, sort=sort_order, axis=alt.Axis(grid=False)),
            x=alt.X(f'{value_field}:Q', title=value_title, axis=alt.Axis(grid=False)),
        )
        width = 620
        height = height or min(560, max(260, 20 * len(data)))
    else:
        encoding = dict(
            x=alt.X(f'{category_field}:N', title=category_title, sort=sort_order, axis=alt.Axis(labelAngle=-30, grid=False)),
            y=alt.Y(f'{value_field}:Q', title=value_title, axis=alt.Axis(grid=False)),
        )
        width = min(680, max(420, 65 * len(data)))
        height = height or 340

    chart = alt.Chart(data).mark_bar(color=color).encode(**encoding, tooltip=tooltip)
    return _finish(chart, title, width, height)


def _decimals_for_width(bin_width):
    """How many decimal places keep a bin label readable, given how wide the bin is."""
    if bin_width >= 5:
        return 0
    if bin_width >= 0.5:
        return 1
    return 2


def _bin_counts(series, bins, min_bins, max_bins):
    """Group a numeric series into readable bins (e.g. '108–434') and count rows per bin."""
    if bins == 'auto':
        edges = np.histogram_bin_edges(series.dropna(), bins='fd')
        bins = int(np.clip(len(edges) - 1, min_bins, max_bins))

    intervals = pd.cut(series, bins=bins, include_lowest=True)
    counts = intervals.value_counts().sort_index().reset_index()
    counts.columns = ['interval', 'Count']

    decimals = _decimals_for_width(counts['interval'].iloc[0].length)
    label = lambda v: f'{v:,.{decimals}f}'
    counts['Bin'] = counts['interval'].apply(lambda iv: f'{label(iv.left)}–{label(iv.right)}')
    return counts[['Bin', 'Count']]


def numeric_distribution(col, bins='auto', discrete=False, min_bins=8, max_bins=20, data=None):
    """Bar chart of how many Records fall into each value (discrete=True) or range of `col`."""
    data = df if data is None else data
    if discrete:
        counts = data[col].value_counts().sort_index().reset_index()
        counts.columns = [col, 'Count']
        plot_data = pd.DataFrame({'Bin': counts[col].astype(str), 'Count': counts['Count']})
    else:
        plot_data = _bin_counts(data[col], bins, min_bins, max_bins)

    return _bar_chart(
        plot_data, 'Bin', 'Count', category_title=col, value_title=f'Number of {UNIT}',
        sort_order=list(plot_data['Bin']), horizontal=False, title=f'Distribution of {col}',
        color=BAR_COLOR, tooltip=[alt.Tooltip('Bin:N', title=col), alt.Tooltip('Count:Q', title=UNIT, format=',')],
    )


def categorical_distribution(col, order=None, top_n=None, horizontal=False, data=None):
    """Bar chart of how many Records fall into each category of `col` (optionally only the top_n)."""
    data = df if data is None else data
    counts = data[col].astype(str).value_counts().reset_index()
    counts.columns = [col, 'Count']
    if top_n is not None:
        counts = counts.head(top_n)
    counts['Share'] = counts['Count'] / len(data)

    sort_order = order if order is not None else ('-x' if horizontal else '-y')
    title = f'Distribution of {col}' if top_n is None else f'Top {top_n} {col} by {UNIT}'
    tooltip = [alt.Tooltip(f'{col}:N'),
               alt.Tooltip('Count:Q', title=UNIT, format=','),
               alt.Tooltip('Share:Q', format='.1%')]

    return _bar_chart(counts, col, 'Count', category_title=col, value_title=f'Number of {UNIT}',
                       sort_order=sort_order, horizontal=horizontal, title=title,
                       color=BAR_COLOR, tooltip=tooltip)
```

#### Link to the Business Objective

> The business objective is to **estimate the outstanding balance of each active loan at the 31 December 2025 reporting date**, which Finance uses to size its expected credit loss (ECL) provisions and to see where lending exposure is concentrated. Each loan in `loans.csv` is linked to an `account_id`, so the cells below check what that account is and how its balance history relates to the loan. Customer-level features are compared with total outstanding balance, the part of that balance not explained by the original loan amount (`residual_after_original`), the worst loan risk grade (A = 1 to E = 5) and the hardship flag. An **active loan** is defined here as one that started on or before 31 December 2025, has not passed the end of its term, and whose linked account is Active.

```python
REPORTING_DATE = pd.Timestamp('2025-12-31')
GRADE_ORDER = {'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5}

loans = pd.read_csv(at.folder_path / "loans.csv", parse_dates=['loan_start_date'])
account_info = pd.read_csv(at.folder_path / "accounts.csv", usecols=['account_id', 'customer_id', 'account_type', 'account_status'])
loans = loans.merge(account_info.drop(columns='customer_id'), on='account_id', how='left')
loan_end_date = loans['loan_start_date'] + pd.to_timedelta(loans['term_months'] * 30.44, unit='D')

started = loans['loan_start_date'] <= REPORTING_DATE
still_in_term = loan_end_date > REPORTING_DATE
active_loans = loans[started & still_in_term & (loans['account_status'] == 'Active')].copy()
active_loans['risk_grade'] = active_loans['loan_risk_grade'].map(GRADE_ORDER)

print('Loans in loans.csv:', len(loans))
print('Start date after the reporting date:', (~started).sum())
print('Past the end of their term but still carrying a balance:', (started & ~still_in_term).sum())
print('Active loans at 31 Dec 2025:', len(active_loans), '| customers:', active_loans['customer_id'].nunique())
print(f"Total outstanding balance: ${active_loans['outstanding_balance'].sum():,.0f}")

term_elapsed = (REPORTING_DATE - active_loans['loan_start_date']).dt.days / (active_loans['term_months'] * 30.44)
largest_tenth = active_loans['outstanding_balance'].nlargest(len(active_loans) // 10).sum()
print('Correlation of outstanding balance with original_amount:',
      round(active_loans['outstanding_balance'].corr(active_loans['original_amount']), 3))
print('Correlation of outstanding / original with share of term elapsed:',
      round((active_loans['outstanding_balance'] / active_loans['original_amount']).corr(term_elapsed), 3))
print(f"Share of outstanding balance in the largest 10% of loans: {largest_tenth / active_loans['outstanding_balance'].sum():.1%}")
print('Share of outstanding balance by loan type:')
(active_loans.groupby('loan_type')['outstanding_balance'].sum() / active_loans['outstanding_balance'].sum()).round(3).sort_values(ascending=False)
```

```python
december = df[df['balance_date'] == '2025-12-31']
linked = active_loans.merge(december[['account_id', 'balance', 'balance_band']], on='account_id', how='left')
offset = linked[linked['account_type'] == 'Offset']

print(f"Active loans whose linked account is in this table: {linked['balance'].notna().mean():.1%}")
print('Linked account types:', linked['account_type'].value_counts().to_dict())
print('Correlation of outstanding balance with the linked account balance (Dec 2025):',
      round(linked['outstanding_balance'].corr(linked['balance']), 3))
print('Offset-linked loans:', len(offset),
      '| median offset balance / outstanding:', round((offset['balance'] / offset['outstanding_balance']).median(), 3),
      f"| fully offset: {(offset['balance'] >= offset['outstanding_balance']).mean():.1%}")
linked.groupby('balance_band')['outstanding_balance'].agg(['count', 'median', 'sum']).round(0)
```

```python
loan_customers = active_loans.groupby('customer_id').agg(
    outstanding_balance=('outstanding_balance', 'sum'),
    original_amount=('original_amount', 'sum'),
    risk_grade=('risk_grade', 'max'),
    hardship=('hardship_program', 'max'),
    vulnerable=('vulnerable_customer_flag', 'max'),
    expected_loss=('expected_loss', 'sum'),
)
loan_history = df[df['customer_id'].isin(loan_customers.index)]
monthly_deposits = loan_history.pivot_table(index='customer_id', columns='balance_date', values='balance', aggfunc='sum')
customer_features = pd.DataFrame({
    'total_deposits': monthly_deposits['2025-12-31'],
    'deposits_12m_change': monthly_deposits['2025-12-31'] - monthly_deposits['2024-12-31'],
    'accounts_held': loan_history[loan_history['balance_date'] == '2025-12-31'].groupby('customer_id').size(),
    'balance_change_volatility': loan_history.groupby('customer_id')['balance_change'].std(),
    'zero_balance_months': loan_history['balance'].eq(0).groupby(loan_history['customer_id']).sum(),
    'stress_base': loan_history.groupby('customer_id')['stress_indicator'].max() % 50,
})
customer_level = loan_customers.join(customer_features)

trend = np.polyfit(customer_level['original_amount'], customer_level['outstanding_balance'], 1)
customer_level['residual_after_original'] = customer_level['outstanding_balance'] - np.polyval(trend, customer_level['original_amount'])

(customer_level.corr()
     .loc[customer_features.columns, ['outstanding_balance', 'residual_after_original', 'risk_grade', 'hardship']]
     .round(3))
```

```python
customer_level.groupby('stress_base').agg(
    customers=('risk_grade', 'size'),
    mean_risk_grade=('risk_grade', 'mean'),
    hardship_share=('hardship', 'mean'),
    vulnerable_share=('vulnerable', 'mean'),
    mean_expected_loss=('expected_loss', 'mean'),
    mean_outstanding=('outstanding_balance', 'mean'),
).round(3)
```

```python
balance_sorted = df.sort_values(['account_id', 'balance_date'])
previous_balance = balance_sorted.groupby('account_id')['balance'].shift()
actual_diff = balance_sorted['balance'] - previous_balance
has_previous = previous_balance.notna()
reconciles = np.isclose(actual_diff[has_previous], balance_sorted.loc[has_previous, 'balance_change'], atol=0.02)
print('Records with a previous month:', has_previous.sum())
print('Share that reconcile with the balance difference:', round(reconciles.mean(), 4))
mismatches = balance_sorted.loc[has_previous][~reconciles]
print('Mismatches:', len(mismatches), '| of which balance is 0:', (mismatches['balance'] == 0).sum())
```

## C. Ethical Consideration of Features


```python
ethical_considerations = """
Purpose and affected parties. Balance history would support the loan-balance and ECL work through deposit context
for each borrower. Directly affected are the 4,300 active-loan customers whose deposit balances, hardship status
and vulnerability status (embedded in `stress_indicator`) would be used, and Finance, risk and the Board who rely
on the provision. Indirectly affected are other holders of the linked accounts (239 active loans are linked to
Youth accounts, which may belong to minors), the remaining customers whose balances are in the table but are not
needed, and shareholders, depositors, auditors and regulators (APRA, ASIC) who rely on accurate provisions.

Hardship and vulnerability. The most serious issue is that `stress_indicator` re-encodes `hardship_program` and
`vulnerable_customer_flag` into a single score. At portfolio level it is legitimate for provisions to reflect that
loans in hardship carry more risk. The danger is function creep: if the same score feeds pricing, limit reductions
or collections, customers would be penalised for asking for help or for being identified as vulnerable (which can
reflect illness, disability, family violence or elder abuse). That would conflict with banks' commitments under
the Banking Code of Practice to support customers in financial difficulty and to take extra care with vulnerable
customers, and could discourage people from seeking hardship assistance. Hiding these inputs inside an opaque
score also makes the model harder to explain and audit.

Fairness and accuracy across segments. Exposure is concentrated among wealthy customers: 85.6% of active
outstanding balance sits on accounts in the Premium or Private Banking bands. A model built mostly on these
customers may be less accurate for lower-balance borrowers, who are also more likely to be financially vulnerable,
so systematic under- or over-provisioning for them would go unnoticed in overall accuracy figures. The synthetic
history adds another risk: accounts opened after 2022 have balances for months before they existed, so features
built on that history would misrepresent real customers.

Aboriginal and Torres Strait Islander peoples. The data has no Indigenous identifier, so outcomes for Aboriginal
and Torres Strait Islander borrowers cannot be measured. Indigenous Australians are more likely to live in remote
areas and to face barriers to banking, so they may be over-represented in the lower-balance segment where the
model is least reliable. Indigenous status should not be inferred from location or balances; any work to
understand impacts should be done with Aboriginal and Torres Strait Islander communities and under Indigenous Data
Sovereignty principles (Maiam nayri Wingara, 2018).

Recommendations. (1) Use `loan_risk_grade`, `hardship_program` and `vulnerable_customer_flag` directly and
transparently in ECL segmentation, not through `stress_indicator`, and restrict access to the vulnerability flag.
(2) Use the model only for portfolio provisioning and exposure reporting, with a documented rule that it is not
used for individual pricing, limits or collections. (3) Limit the data to accounts linked to active loans (APP 3
minimisation) and mask months before each account was opened. (4) Report prediction errors by balance band and
hardship status, not just overall. (5) Treat offset balances as an exposure measure for reporting only; using the
model to trigger set-off against customer deposits would be a separate decision requiring its own legal review.

References: AASB 9 Financial Instruments (section 5.5); APRA Prudential Standard APS 220 Credit Risk Management;
Australian Banking Association, Banking Code of Practice (financial difficulty and vulnerable customers); Privacy
Act 1988 (Cth), Australian Privacy Principles 3 and 6; Maiam nayri Wingara Indigenous Data Sovereignty Principles
(2018); Australia's AI Ethics Principles (Department of Industry, Science, Energy and Resources, 2019).
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='ethical_considerations', value=ethical_considerations)
```
