```python
# <Student to fill this section>
dataset_insights = """
`economics.csv` contains 72 rows and 11 columns, one row per month from January 2020 to December 2025, with no
gaps and no missing values. `economic_date` is the month, eight columns are continuous indicators
(`inflation_rate`, `interest_rate`, `unemployment_rate`, `consumer_sentiment`, `loan_stress_index`,
`spending_factor`, `housing_price_index`, `equity_market_index`), `economic_cycle` is a four-phase label and
`recession_flag` is binary.

The business objective is to estimate the outstanding balance of each active loan at 31 December 2025 so Finance
can size expected credit loss (ECL) provisions and see where exposure is concentrated. Economic data matters for
ECL in principle, because AASB 9 requires provisions to reflect forecasts of future economic conditions, not just
past losses. However, this table has no customer or loan key, so it can only be joined by month: to `loans.csv` by
the month a loan started, or to `loan_repayments.csv` by payment month. Only 1,828 of the 4,872 active loans
(29.8% of outstanding balance) started within 2020–2025, so origination conditions are unknown for the other 70%,
and at the reporting date every loan shares the same single set of December 2025 values.

The loan-level checks at the start of "Link to the Business Objective" frame what any table can add. Of the 10,673
loans, 1,813 start after the reporting date and 3,422 are past the end of their term yet still carry a balance, so
the active population (4,872 loans, 4,300 customers, $2.13bn outstanding) depends on how "active" is defined, and
Finance should confirm that definition before the provision is sized. Outstanding balance is driven mainly by
`original_amount` (r = 0.835), while the share still owed is unrelated to how much of the term has passed (r =
0.024): balances do not amortise in this data, so the variation left after loan size behaves like noise that no
behavioural table can explain. Exposure is concentrated, with home loans holding 75.1% and investment loans 22.5%
of the balance, and the largest 10% of loans holding 32.9%.

The evidence in "Link to the Business Objective" shows no usable relationship. At origination, no indicator has
|r| above 0.123 with outstanding balance, and the largest values (`housing_price_index` -0.123, `interest_rate`
-0.110) reflect which loans are still active rather than the economy: older active loans are larger (mean
outstanding $457k for loans started in Expansion vs $292k in Recovery). Loan interest rates are unrelated to the
economic `interest_rate` (r = 0.022), and late-payment rates stay flat at 8.5–9.0% a year, moving in the opposite
direction to `loan_stress_index` (r = -0.40). The economy was not used to generate borrower behaviour in this
data.

The table is also a scripted scenario rather than observed history: four 18-month phases with no Contraction,
`recession_flag` always 0, inflation and interest rates that miss Australia's 2022–2023 surge, and strong
collinearity (`consumer_sentiment` and `spending_factor` r = 0.997). It should not be a model input for the
balance estimate. Its realistic role is as a documented scenario overlay for ECL (base and downside settings
applied on top of a model), with the downside built from external forecasts because this data contains no
downturn.
"""
```

### Helper Functions

```python
import numpy as np

UNIT = 'Months'
BAR_COLOR = '#2563eb'
AXIS_CFG = dict(domainColor='#9ca3af', tickColor='#9ca3af', labelColor='#000', titleColor='#000',
                labelFontSize=12, titleFontSize=13, labelLimit=160)


def _finish(chart, title, width, height):
    """Apply the title, size and axis styling shared by every chart in this notebook."""
    return (chart.properties(title=alt.TitleParams(title, fontSize=15, anchor='start'), width=width, height=height)
                 .configure_view(strokeWidth=0)
                 .configure_axis(**AXIS_CFG))


def _bar_chart(data, category_field, value_field, *, category_title, value_title,
               sort_order, tooltip, color, horizontal, title, height=None):
    """One bar per category, laid out as columns, or as rows when horizontal=True. Width and height
    are capped so the chart always fits on screen without scrolling."""
    if horizontal:
        encoding = dict(
            y=alt.Y(f'{category_field}:N', title=category_title, sort=sort_order, axis=alt.Axis(grid=False)),
            x=alt.X(f'{value_field}:Q', title=value_title, axis=alt.Axis(grid=False)),
        )
        width = 620
        height = height or min(560, max(260, 20 * len(data)))
    else:
        encoding = dict(
            x=alt.X(f'{category_field}:N', title=category_title, sort=sort_order, axis=alt.Axis(labelAngle=-30, grid=False)),
            y=alt.Y(f'{value_field}:Q', title=value_title, axis=alt.Axis(grid=False)),
        )
        width = min(680, max(420, 65 * len(data)))
        height = height or 340

    chart = alt.Chart(data).mark_bar(color=color).encode(**encoding, tooltip=tooltip)
    return _finish(chart, title, width, height)


def _decimals_for_width(bin_width):
    """How many decimal places keep a bin label readable, given how wide the bin is."""
    if bin_width >= 5:
        return 0
    if bin_width >= 0.5:
        return 1
    return 2


def _bin_counts(series, bins, min_bins, max_bins):
    """Group a numeric series into readable bins (e.g. '108–434') and count rows per bin."""
    if bins == 'auto':
        edges = np.histogram_bin_edges(series.dropna(), bins='fd')
        bins = int(np.clip(len(edges) - 1, min_bins, max_bins))

    intervals = pd.cut(series, bins=bins, include_lowest=True)
    counts = intervals.value_counts().sort_index().reset_index()
    counts.columns = ['interval', 'Count']

    decimals = _decimals_for_width(counts['interval'].iloc[0].length)
    label = lambda v: f'{v:,.{decimals}f}'
    counts['Bin'] = counts['interval'].apply(lambda iv: f'{label(iv.left)}–{label(iv.right)}')
    return counts[['Bin', 'Count']]


def numeric_distribution(col, bins='auto', discrete=False, min_bins=8, max_bins=20, data=None):
    """Bar chart of how many Months fall into each value (discrete=True) or range of `col`."""
    data = df if data is None else data
    if discrete:
        counts = data[col].value_counts().sort_index().reset_index()
        counts.columns = [col, 'Count']
        plot_data = pd.DataFrame({'Bin': counts[col].astype(str), 'Count': counts['Count']})
    else:
        plot_data = _bin_counts(data[col], bins, min_bins, max_bins)

    return _bar_chart(
        plot_data, 'Bin', 'Count', category_title=col, value_title=f'Number of {UNIT}',
        sort_order=list(plot_data['Bin']), horizontal=False, title=f'Distribution of {col}',
        color=BAR_COLOR, tooltip=[alt.Tooltip('Bin:N', title=col), alt.Tooltip('Count:Q', title=UNIT, format=',')],
    )


def categorical_distribution(col, order=None, top_n=None, horizontal=False, data=None):
    """Bar chart of how many Months fall into each category of `col` (optionally only the top_n)."""
    data = df if data is None else data
    counts = data[col].astype(str).value_counts().reset_index()
    counts.columns = [col, 'Count']
    if top_n is not None:
        counts = counts.head(top_n)
    counts['Share'] = counts['Count'] / len(data)

    sort_order = order if order is not None else ('-x' if horizontal else '-y')
    title = f'Distribution of {col}' if top_n is None else f'Top {top_n} {col} by {UNIT}'
    tooltip = [alt.Tooltip(f'{col}:N'),
               alt.Tooltip('Count:Q', title=UNIT, format=','),
               alt.Tooltip('Share:Q', format='.1%')]

    return _bar_chart(counts, col, 'Count', category_title=col, value_title=f'Number of {UNIT}',
                       sort_order=sort_order, horizontal=horizontal, title=title,
                       color=BAR_COLOR, tooltip=tooltip)


LINE_COLOR = '#9ca3af'
CYCLE_ORDER = ['Expansion', 'Peak', 'Slowdown', 'Recovery']


def time_series(col, data=None):
    """Monthly line of `col`, with points coloured by economic_cycle to show regime shifts."""
    data = df if data is None else data
    plot_data = data[['economic_date', 'economic_cycle', col]].copy()
    plot_data['economic_date'] = pd.to_datetime(plot_data['economic_date'])

    base = alt.Chart(plot_data).encode(
        x=alt.X('economic_date:T', title='Month'),
        y=alt.Y(f'{col}:Q', title=col, scale=alt.Scale(zero=False)),
    )
    line = base.mark_line(color=LINE_COLOR)
    points = base.mark_circle(size=45, opacity=1).encode(
        color=alt.Color('economic_cycle:N', title='Economic cycle', sort=CYCLE_ORDER),
        tooltip=[alt.Tooltip('economic_date:T', format='%b %Y'), 'economic_cycle:N', alt.Tooltip(f'{col}:Q')],
    )
    return _finish(line + points, f'{col} over Time', 680, 320)


def cycle_summary(col, data=None):
    """Mean, min and max of `col` within each economic cycle, in chronological order."""
    data = df if data is None else data
    return data.groupby('economic_cycle')[col].agg(['mean', 'min', 'max']).round(3).reindex(CYCLE_ORDER)
```

#### Link to the Business Objective

> The business objective is to **estimate the outstanding balance of each active loan at the 31 December 2025 reporting date**, which Finance uses to size its expected credit loss (ECL) provisions and to see where lending exposure is concentrated. Economic data has no customer or loan key, so the cells below join it to `loans.csv` by the month each loan started (origination conditions) and to `loan_repayments.csv` by payment month (repayment behaviour). An **active loan** is defined here as one that started on or before 31 December 2025, has not passed the end of its term, and whose linked account is Active.

```python
REPORTING_DATE = pd.Timestamp('2025-12-31')
GRADE_ORDER = {'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5}

loans = pd.read_csv(at.folder_path / "loans.csv", parse_dates=['loan_start_date'])
account_info = pd.read_csv(at.folder_path / "accounts.csv", usecols=['account_id', 'customer_id', 'account_type', 'account_status'])
loans = loans.merge(account_info.drop(columns='customer_id'), on='account_id', how='left')
loan_end_date = loans['loan_start_date'] + pd.to_timedelta(loans['term_months'] * 30.44, unit='D')

started = loans['loan_start_date'] <= REPORTING_DATE
still_in_term = loan_end_date > REPORTING_DATE
active_loans = loans[started & still_in_term & (loans['account_status'] == 'Active')].copy()
active_loans['risk_grade'] = active_loans['loan_risk_grade'].map(GRADE_ORDER)

print('Loans in loans.csv:', len(loans))
print('Start date after the reporting date:', (~started).sum())
print('Past the end of their term but still carrying a balance:', (started & ~still_in_term).sum())
print('Active loans at 31 Dec 2025:', len(active_loans), '| customers:', active_loans['customer_id'].nunique())
print(f"Total outstanding balance: ${active_loans['outstanding_balance'].sum():,.0f}")

term_elapsed = (REPORTING_DATE - active_loans['loan_start_date']).dt.days / (active_loans['term_months'] * 30.44)
largest_tenth = active_loans['outstanding_balance'].nlargest(len(active_loans) // 10).sum()
print('Correlation of outstanding balance with original_amount:',
      round(active_loans['outstanding_balance'].corr(active_loans['original_amount']), 3))
print('Correlation of outstanding / original with share of term elapsed:',
      round((active_loans['outstanding_balance'] / active_loans['original_amount']).corr(term_elapsed), 3))
print(f"Share of outstanding balance in the largest 10% of loans: {largest_tenth / active_loans['outstanding_balance'].sum():.1%}")
print('Share of outstanding balance by loan type:')
(active_loans.groupby('loan_type')['outstanding_balance'].sum() / active_loans['outstanding_balance'].sum()).round(3).sort_values(ascending=False)
```

```python
INDICATORS = ['inflation_rate', 'interest_rate', 'unemployment_rate', 'consumer_sentiment',
              'loan_stress_index', 'spending_factor', 'housing_price_index', 'equity_market_index']
economy = df.assign(month=pd.to_datetime(df['economic_date']).dt.to_period('M'))
originated = (active_loans.assign(month=active_loans['loan_start_date'].dt.to_period('M'))
                          .merge(economy, on='month', suffixes=('_loan', '')))

print('Active loans originated between Jan 2020 and Dec 2025:', len(originated), 'of', len(active_loans),
      f"({originated['outstanding_balance'].sum() / active_loans['outstanding_balance'].sum():.1%} of outstanding balance)")
(originated[INDICATORS + ['outstanding_balance', 'original_amount', 'risk_grade', 'interest_rate_loan']].corr()
     .loc[INDICATORS, ['outstanding_balance', 'original_amount', 'risk_grade', 'interest_rate_loan']]
     .round(3))
```

```python
originated.groupby('economic_cycle').agg(
    loans=('loan_id', 'size'),
    outstanding_balance=('outstanding_balance', 'sum'),
    mean_outstanding=('outstanding_balance', 'mean'),
    grade_c_or_worse=('risk_grade', lambda s: (s >= 3).mean()),
).reindex(CYCLE_ORDER).round(3)
```

```python
repayments = pd.read_csv(at.folder_path / "loan_repayments.csv",
                         usecols=['payment_date', 'late_payment', 'missed_payment'], parse_dates=['payment_date'])
monthly_repayments = (repayments.assign(month=repayments['payment_date'].dt.to_period('M'))
                                .groupby('month')[['late_payment', 'missed_payment']].mean())
repayment_vs_economy = economy.set_index('month').join(monthly_repayments, how='inner')

print(f"Repayment records dated after the reporting date: {(repayments['payment_date'] > REPORTING_DATE).mean():.1%}")
print('Late-payment rate by year:', repayment_vs_economy.groupby(repayment_vs_economy.index.year)['late_payment'].mean().round(4).to_dict())
repayment_vs_economy[INDICATORS + ['late_payment', 'missed_payment']].corr().loc[INDICATORS, ['late_payment', 'missed_payment']].round(3)
```

## C. Ethical Consideration of Features


```python
ethical_considerations = """
Purpose and affected parties. Economic indicators contain no personal information, so the main ethical risks here
are about honesty, accuracy and the consequences of the provision rather than privacy. Directly affected are the
CFO, Finance and the Board audit committee, who sign off the provision, and the 4,300 active-loan borrowers, whose
loans are provisioned. Indirectly affected are shareholders and investors (provisions change reported profit),
depositors (provisions protect the bank's capital), external auditors and APRA (who review whether provisions are
supportable), and future borrowers, because a larger provision often leads a bank to tighten lending in the
segments or regions it considers riskier.

Honest representation. AASB 9 requires ECL to reflect reasonable and supportable forecasts of future economic
conditions. This series is a scripted scenario: no recession, inflation and interest rates that miss Australia's
2022–2023 experience, and no measurable link to loan balances, risk grades or repayment behaviour (late payments
even move the opposite way to `loan_stress_index`, r = -0.40). Presenting macro "effects" from this data to the
Board, auditors or APRA would overstate what the model knows and could misstate the provision. The data's
limitations should be stated explicitly, and economic scenarios should come from credible external sources.

Fairness of macro features. Because every loan shares the same economic values in a given month, using conditions
at origination as a model input would give borrowers higher or lower predicted risk purely because of when they
borrowed (e.g. during the Slowdown), not because of anything about them. That may be acceptable for a
portfolio-level overlay, but not for decisions about individual customers. A related risk is leakage: 45.6% of
repayment records are dated after the reporting date, and using them would make the model look more accurate than
it can be in practice.

Regional and Aboriginal and Torres Strait Islander communities. National indices hide large regional differences.
Remote and regional areas, including many Aboriginal and Torres Strait Islander communities, have different
housing markets, employment conditions and access to credit, so a national scenario can under- or over-state risk
there. If a downside scenario leads the bank to restrict lending by region, these communities could lose access to
credit without evidence that their risk changed. Regional and Indigenous-specific economic data should only be
introduced with community engagement and under Indigenous Data Sovereignty principles (Maiam nayri Wingara, 2018).

Recommendations. (1) Do not use this table as a model input for outstanding balances. (2) Build ECL scenarios
(base, downside, upside) from external forecasts such as RBA and Treasury publications, with probability weights
approved by Finance and disclosed to auditors. (3) Filter all behavioural data to the reporting date to avoid
leakage. (4) Before any lending-policy change based on a scenario, assess its impact on regional, remote and
Indigenous communities. (5) Document assumptions and limitations so the provision can be explained and challenged,
consistent with Australia's AI Ethics Principles.

References: AASB 9 Financial Instruments (paragraph 5.5.17, reasonable and supportable forecasts); APRA Prudential
Standard APS 220 Credit Risk Management; Reserve Bank of Australia, Statement on Monetary Policy (forecast
source); Maiam nayri Wingara Indigenous Data Sovereignty Principles (2018); Australia's AI Ethics Principles
(Department of Industry, Science, Energy and Resources, 2019).
"""
```

```python
# DO NOT MODIFY THE CODE IN THIS CELL
print_tile(size="h3", key='ethical_considerations', value=ethical_considerations)
```
