### Helper Functions

```python
# The Helper Functions are AI generated to assist in graphical representations of band ranges
# between numerical and categorical variables (adapted from the AT2 EDA notebook).
# All aggregation is done in pandas first so Altair only receives the summarised counts,
# which keeps charts fast on 1.5M transactions and avoids Altair's 5,000 row limit.

import numpy as np

AXIS_CFG = dict(domainColor='#9ca3af', tickColor='#9ca3af', labelColor='#000', titleColor='#000')


def numeric_distribution(col, bins='auto', discrete=False, min_bins=8, max_bins=20, data=None):
    data = df if data is None else data
    if discrete:
        counts = data[col].value_counts().sort_index().reset_index()
        counts.columns = [col, 'Count']
        counts['Bin'] = counts[col].astype(str)
    else:
        if bins == 'auto':
            # Freedman-Diaconis bin width, clipped to [min_bins, max_bins] so labels stay readable
            edges = np.histogram_bin_edges(data[col].dropna(), bins='fd')
            n_bins = int(np.clip(len(edges) - 1, min_bins, max_bins))
        else:
            n_bins = bins

        band = pd.cut(data[col], bins=n_bins, include_lowest=True)
        counts = band.value_counts().sort_index().reset_index()
        counts.columns = ['BandCat', 'Count']

        band_width = counts['BandCat'].iloc[0].length
        if band_width >= 5:
            fmt = lambda v: f'{v:,.0f}'
        elif band_width >= 0.5:
            fmt = lambda v: f'{v:,.1f}'
        else:
            fmt = lambda v: f'{v:,.2f}'

        counts['Bin'] = counts['BandCat'].apply(lambda iv: f'{fmt(iv.left)}–{fmt(iv.right)}')

    sort_order = list(counts['Bin'])
    plot_data = counts[['Bin', 'Count']].copy()

    chart = alt.Chart(plot_data).mark_bar(color='#2563eb').encode(
        x=alt.X('Bin:N', title=col, sort=sort_order, axis=alt.Axis(labelAngle=-30, grid=False)),
        y=alt.Y('Count:Q', title='Number of Transactions', axis=alt.Axis(grid=False)),
        tooltip=[alt.Tooltip('Bin:N', title=col), alt.Tooltip('Count:Q', title='Transactions', format=',')]
    ).properties(
        title=f'Distribution of {col}', width=520, height=350
    ).configure_view(strokeWidth=0
    ).configure_axis(**AXIS_CFG)

    return chart


def categorical_distribution(col, order=None, top_n=None, horizontal=False, data=None):
    data = df if data is None else data
    counts = data[col].astype(str).value_counts().reset_index()
    counts.columns = [col, 'Count']
    if top_n is not None:
        counts = counts.head(top_n)
    counts['Share'] = counts['Count'] / len(data)
    sort_order = order if order is not None else ('-x' if horizontal else '-y')
    title = f'Distribution of {col}' if top_n is None else f'Top {top_n} {col} by Transactions'

    tooltip = [alt.Tooltip(f'{col}:N'),
               alt.Tooltip('Count:Q', title='Transactions', format=','),
               alt.Tooltip('Share:Q', format='.1%')]

    if horizontal:
        enc = dict(
            y=alt.Y(f'{col}:N', title=col, sort=sort_order, axis=alt.Axis(grid=False)),
            x=alt.X('Count:Q', title='Number of Transactions', axis=alt.Axis(grid=False)),
        )
        size = dict(width=520, height=max(250, 18 * len(counts)))
    else:
        enc = dict(
            x=alt.X(f'{col}:N', title=col, sort=sort_order, axis=alt.Axis(labelAngle=0, grid=False)),
            y=alt.Y('Count:Q', title='Number of Transactions', axis=alt.Axis(grid=False)),
        )
        size = dict(width=max(440, 70 * len(counts)), height=350)

    chart = alt.Chart(counts).mark_bar(color='#2563eb').encode(
        **enc, tooltip=tooltip
    ).properties(title=title, **size
    ).configure_view(strokeWidth=0
    ).configure_axis(**AXIS_CFG)

    return chart


def group_mean_chart(group_col, value_col='aml_risk', order=None, data=None):
    """Mean of value_col for each level of group_col, used to compare risk across segments."""
    data = df if data is None else data
    agg = data.groupby(group_col)[value_col].agg(['mean', 'count']).reset_index()
    agg[group_col] = agg[group_col].astype(str)
    sort_order = order if order is not None else '-y'

    chart = alt.Chart(agg).mark_bar(color='#0f766e').encode(
        x=alt.X(f'{group_col}:N', title=group_col, sort=sort_order, axis=alt.Axis(labelAngle=0, grid=False)),
        y=alt.Y('mean:Q', title=f'Mean {value_col}', axis=alt.Axis(grid=False)),
        tooltip=[alt.Tooltip(f'{group_col}:N'),
                 alt.Tooltip('mean:Q', title=f'Mean {value_col}', format='.3f'),
                 alt.Tooltip('count:Q', title='Transactions', format=',')]
    ).properties(
        title=f'Mean {value_col} by {group_col}', width=max(440, 70 * len(agg)), height=300
    ).configure_view(strokeWidth=0
    ).configure_axis(**AXIS_CFG)

    return chart
```
